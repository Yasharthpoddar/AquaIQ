```python
import matplotlib.pyplot as plt

districts = [
    'Cuddalore',
    'Tiruppur',
    'Sivaganga',
    'Murshidabad',
    'Karur'
]

for district in districts:

    temp = df[df['district_name'] == district].copy()

    # Sort by date
    temp = temp.sort_values('date')

    plt.figure(figsize=(12, 5))

    plt.plot(
        temp['date'],
        temp['currentlevel']
    )

    plt.title(f'Groundwater Level Trend - {district}')
    plt.xlabel('Date')
    plt.ylabel('Groundwater Level')
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

districts = [
    'Cuddalore',
    'Tiruppur',
    'Sivaganga',
    'Murshidabad',
    'Karur'
]

# Make sure date is datetime
df['date'] = pd.to_datetime(df['date'])

# Create a copy for our plots
plot_df = df[df['district_name'].isin(districts)].copy()

# Create Year-Month
plot_df['month'] = plot_df['date'].dt.to_period('M')

# Monthly average groundwater level
monthly_gwl = (
    plot_df
    .groupby(['district_name', 'month'])['currentlevel']
    .mean()
    .reset_index()
)

monthly_gwl['month'] = monthly_gwl['month'].dt.to_timestamp()

# Create 5 separate graphs
for district in districts:

    temp = monthly_gwl[
        monthly_gwl['district_name'] == district
    ]

    plt.figure(figsize=(12, 5))

    plt.plot(
        temp['month'],
        temp['currentlevel'],
        linewidth=1.5
    )

    plt.title(f'Groundwater Level Trend (2013–2023) - {district}')
    plt.xlabel('Year')
    plt.ylabel('Average Groundwater Level')
    plt.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()
```
